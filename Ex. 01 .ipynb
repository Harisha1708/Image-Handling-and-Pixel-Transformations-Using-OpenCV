10. Read in the image ('Apollo-11-launch.jpg').

```python
img = cv2.imread('Apollo-11-launch.jpg')
```

11. Add the following text to the dark area at the bottom of the image (centered on the image):

```python
text = 'Apollo 11 Saturn V Launch, July 16, 1969'
font_face = cv2.FONT_HERSHEY_PLAIN

# Get image dimensions
h, w = img.shape[:2]

# Get text size
(text_width, text_height), _ = cv2.getTextSize(text, font_face, 2, 2)

# Calculate center position
x = (w - text_width) // 2
y = h - 20

# Put text on image
cv2.putText(img, text, (x, y), font_face, 2, (255, 255, 255), 2)

# Display image only
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.title("Image with Text")
plt.axis('off')
plt.show()
```

14. Read the image ('boy.jpg').

```python
img = cv2.imread('boy.jpg')
```

15. Adjust the brightness of the image.

```python
# Create a matrix of ones (with data type float64)
# matrix_ones = 
# YOUR CODE HERE
matrix_ones = np.ones(img.shape, dtype="float64") * 50

# Increase brightness
img_brighter = cv2.add(img, matrix_ones.astype('uint8'))
```

16. Create brighter and darker images.

```python
#img_brighter = cv2.add(img, matrix)
#img_darker = cv2.subtract(img, matrix)
# YOUR CODE HERE
matrix = np.ones(img.shape, dtype="uint8") * 50

# Brighter image
img_brighter = cv2.add(img, matrix)

# Darker image
img_darker = cv2.subtract(img, matrix)
```

17. Display the images (Original Image, Darker Image, Brighter Image).

```python
img_rgb  = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
bright_rgb = cv2.cvtColor(img_brighter, cv2.COLOR_BGR2RGB)
dark_rgb = cv2.cvtColor(img_darker, cv2.COLOR_BGR2RGB)

# Original Image
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis('off')
plt.show()

# Brighter Image
plt.imshow(bright_rgb)
plt.title("Brighter Image")
plt.axis('off')
plt.show()

# Darker Image
plt.imshow(dark_rgb)
plt.title("Darker Image")
plt.axis('off')
plt.show()
```

18. Modify the image contrast.

```python
# Create two higher contrast images using the 'scale' option with factors of 1.1 and 1.2 (without overflow fix)
# matrix1 = 
# matrix2 = 
# img_higher1 = 
# img_higher2 = 
# YOUR CODE HERE
matrix1 = np.ones(img.shape, dtype="float64") * 1.1
matrix2 = np.ones(img.shape, dtype="float64") * 1.2

# Apply scaling (no overflow handling)
img_higher1 = (img * matrix1).astype('uint8')
img_higher2 = (img * matrix2).astype('uint8')
```

19. Display the images (Original, Lower Contrast, Higher Contrast).

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
low_rgb = cv2.cvtColor(img_higher1, cv2.COLOR_BGR2RGB)   # 1.1
high_rgb = cv2.cvtColor(img_higher2, cv2.COLOR_BGR2RGB)  # 1.2

# Original Image
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis('off')
plt.show()

# Lower Contrast Image
plt.imshow(low_rgb)
plt.title("Lower Contrast (1.1)")
plt.axis('off')
plt.show()

# Higher Contrast Image
plt.imshow(high_rgb)
plt.title("Higher Contrast (1.1)")
plt.axis('off')
plt.show()
```

20. Split the image (boy.jpg) into the B,G,R components & Display the channels.

```python
b, g, r = cv2.split(img)
```

21. _merged the R, G, B , display along with orginal image 

```python
# Blue Channel
plt.imshow(b, cmap='gray')
plt.title("Blue Channel")
plt.axis('off')
plt.show()

# Green Channel
plt.imshow(g, cmap='gray')
plt.title("Green Channel")
plt.axis('off')
plt.show()

# Red Channel
plt.imshow(r, cmap='gray')
plt.title("Red Channel")
plt.axis('off')
plt.show()
```

```python
merged_img = cv2.merge((b, g, r))

# Convert to RGB for display
merged_rgb = cv2.cvtColor(merged_img, cv2.COLOR_BGR2RGB)
original_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Original Image
plt.imshow(original_rgb)
plt.title("Original Image")
plt.axis('off')
plt.show()

# Merged Image
plt.imshow(merged_rgb)
plt.title("Merged RGB Image")
plt.axis('off')
plt.show()
```

22. Split the image into the H, S, V components &  Display the channels.

```python
# Convert BGR image to HSV
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# Split HSV channels
h, s, v = cv2.split(hsv)

# Display Hue channel
plt.imshow(h, cmap='gray')
plt.title("Hue Channel")
plt.axis('off')
plt.show()

# Display Saturation channel
plt.imshow(s, cmap='gray')
plt.title("Saturation Channel")
plt.axis('off')
plt.show()

# Display Value channel
plt.imshow(v, cmap='gray')
plt.title("Value Channel")
plt.axis('off')
plt.show()
```

23. _merged the H, S, V, display along with orginal image 

```python
# Merge HSV channels
merged_hsv = cv2.merge((h, s, v))

# Convert HSV back to BGR
bgr_img = cv2.cvtColor(merged_hsv, cv2.COLOR_HSV2BGR)

# Convert BGR to RGB for matplotlib
rgb_img = cv2.cvtColor(bgr_img, cv2.COLOR_BGR2RGB)
original_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Display Original Image
plt.imshow(original_rgb)
plt.title("Original Image")
plt.axis('off')
plt.show()

# Display Merged HSV Image
plt.imshow(rgb_img)
plt.title("Merged HSV Image")
plt.axis('off')
plt.show()
```
